#### Vector-vector multiplication

##### Vector-vector multiplication is also called the dot product, or sometimes the inner product.

```python
print(u)
print(v)
u * v
```

##### Implementing the vector sum in a function

```python
def vector_vector_multiplication(u, v):
    assert u.shape[0] == v.shape[0]

    n = u.shape[0]
    result = 0
    for i in range(n):
        result = result + u[i] * v[i]
    return result
    
print(vector_vector_multiplication(u, v))
```

##### There is a defined function/operation in numpy for this called the dot operation

```python
u.dot(v)
```

#### Matrix-vector multiplication

```python
U = np.array([[2,4,5,6],
             [1,2,1,2],
             [3,1,2,1]])
U.shape
```

##### Python implementation 

```python
v = np.array([1, 0.5, 2, 1])
def matrix_vector_multiplication(U, v):
    assert U.shape[1] == v.shape[0]

    n=U.shape[0]
    result = np.zeros(n)
    for i in range(n):
        result[i] = vector_vector_multiplication(U[i],v)
    return result

print(U)
print(v)
matrix_vector_multiplication(U,v)
```

##### using numpy dot operation

```python
U.dot(v)
```

#### Matrix-matrix multiplication

```python
print(U)
```

```python
V = np.array([[1, 1, 2],[0, 0.5, 1],[0, 2, 1],[2, 1, 0]])
print(V)
```

##### Python implementation

```python
def matrix_matrix_multiplication(U, V):
    assert U.shape[1] == V.shape[0]

    rows = U.shape[0]
    cols = V.shape[1]

    result = np.zeros((rows, cols))
    
    for i in range(cols):
        vi = V[:, i]
        Uvi = matrix_vector_multiplication(U,vi)
        result[:, i] = Uvi

    return result

matrix_matrix_multiplication(U, V)
```

##### using numpy operation - dot()

```python
U.dot(V)
```

##### This way we expressed matrix-matrix multiplication using matrix-vector multiplication, and in turn we expressed matrix-vector multiplication with vector-vector multiplication.


### Identity matrix

###### The identity matrix is a square matrix where on the diagonal we have ones, and zeros everywhere else. 

###### The identity matrix is like the number one: if we multiply any number by one, we get the number back, no matter on which side. This is the same thing, but for matrices. Why is it useful? When we take any matrix - say U - and multiply it by I (of course, the dimensions should match), we get U back. And it doesn't matter on which side we put it - we still get U back.

```python
import numpy as np
```

```python
I = np.eye(3)
I
```

```python
V = np.array([[1, 1, 2],[0, 0.5, 1],[0, 2, 1],[2, 1, 0]])
V
```

```python
V.dot(I) ## We multiplied V by I and got V back.
```

#### Inverse

##### The identity matrix is useful for explaining what a matrix inverse is. Let's say we have a matrix A. The inverse of A, usually written A⁻¹, is a matrix such that when we multiply it by A, we get I.

###### Only square matrices - matrices where the number of rows equals the number of columns - have an inverse

```python
Vs = V[[0, 1, 2]]
Vs
```

```python
Vs_inv = np.linalg.inv(Vs)
Vs_inv
```

```python
result = Vs_inv.dot(Vs)
result
```

##### This will be quite useful for linear regression
